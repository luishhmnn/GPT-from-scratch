## Introduction to PyTorch
This playbook demonstrates basic PyTorch functionalities.

```python
import torch
```

### Tensors

```python
# Create first 2D tensor
tensor_2d_a = torch.tensor([[1, 2], [3, 4]])
print(tensor_2d_a)
```

```python
# Create second 2D tensor
tensor_2d_b = torch.tensor([[4, 3], [2, 1]])
print(tensor_2d_b)
```

```python
# Matrix multiplication
result = torch.matmul(tensor_2d_a, tensor_2d_b)
print(result)
```

```python
# Transposing the first tensor
transposed_a_transposed = tensor_2d_a.T
print(transposed_a_transposed)
```

```python
# Create a 1-D tensor of evenly spaced values within a given interval
print(torch.arange(start=0, end=4, step=1))
print(torch.arange(4)) # Default start is 0 and step is 1
```

```python
# Creating embeddings
embeddings = torch.nn.Embedding(num_embeddings=6, embedding_dim=4)
print(embeddings.weight)
```
